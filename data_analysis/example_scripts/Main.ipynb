* For the 300 bp segments that would be the structure factor for rod-like WLC: $S(q)_{rod}=\frac{2N}{v^2}\left[v Si(\nu) + cos(v) - 1\right];\ (v=Lq)$
* For the linearized pICOz that would be the Debye structure factor for linear Gaussian chains: $S(q)_{Debye} = \frac{2N}{(R_gq)^2}\left[(R_gq)^4 - 1 + e^{-(R_gq)^2}\right]$
